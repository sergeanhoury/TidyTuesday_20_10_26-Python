```python
data=salary[['median_starting_salary','ai_exposure']].copy()
```

```python
data.head()
```

```python
data["ai_exposure"].describe()
```

```python
#Scale data to standardize values
data = ((data - data.min()) / (data.max() - data.min())) * 10 + 1
```

```python
data.describe()
```

```python
import matplotlib.pyplot as plt
x=data['median_starting_salary']
y=data['ai_exposure']
plt=plt.scatter(x,y)
```

```python
from sklearn.cluster import KMeans
km = KMeans(n_clusters=3)
y_predicted = km.fit_predict(data[['median_starting_salary','ai_exposure']])
y_predicted
```

```python
data['cluster']=y_predicted
data.head()
```

```python
import matplotlib.pyplot as plt
df1 = data[data.cluster==0]
df2 = data[data.cluster==1]
df3 = data[data.cluster==2]
plt.scatter(df1["median_starting_salary"],df1['ai_exposure'],color='green')
plt.scatter(df2["median_starting_salary"],df2['ai_exposure'],color='red')
plt.scatter(df3["median_starting_salary"],df3['ai_exposure'],color='black')
plt.scatter(km.cluster_centers_[:,0],km.cluster_centers_[:,1],color='purple',marker='*',label='centroid')
plt.xlabel('Salary')
plt.ylabel('ai exposure')
plt.legend()
```

```python
#Elbow Plot

sse = []
k_rng = range(1,10)
for k in k_rng:
    km = KMeans(n_clusters=k)
    km.fit(data[['median_starting_salary','ai_exposure']])
    sse.append(km.inertia_)
```

```python
plt.xlabel('K')
plt.ylabel('Sum of squared error')
plt.plot(k_rng,sse)
```
